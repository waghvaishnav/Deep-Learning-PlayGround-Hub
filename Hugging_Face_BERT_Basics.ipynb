```python
# from transformer in huggingface importing bert and tokenizer
from transformers import BertTokenizer, BertModel

# importing the bert base model
tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")
```

```python
tokens = tokenizer("i love India",return_tensor="pt",max_length=3)
tokens
```

```python
tokens = tokenizer(["Python Machine Learning","Artificial Intelligence and Machine Learning"],return_tensor="pt",padding=True)
tokens
```

```python
tokens = tokenizer(["Python Machine Learning","Artificial Intelligence and Machine Learning"],return_tensors="pt",padding="max_length",max_length=4,truncation=True)
tokens
```

# Make Prediction

```python
model = BertModel.from_pretrained("bert-base-uncased")

output = model(**tokens)
```

```python
# contexual embedding for the sentances ("the first embedding can captures meaning of entire sentence")

output["last_hidden_state"]
```

## pooler representing the entire sentence meaning


```python
# pooler representing the entire sentence

output["pooler_output"]
```
